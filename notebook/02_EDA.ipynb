## Finding Best Features 

```python
corr=df[num_cols].corr()
corr['churned'].abs().sort_values(ascending=False)
```

```python
top_num_cols=['days_since_last_order', 'predicted_churn_prob', 'nps_score',
       'support_tickets_30d', 'complaint_count', 'satisfaction_score',
       'total_spend', 'avg_order_value', 'monthly_spend', 'pages_per_session',
       'tenure_months']
```

```python
from scipy.stats import chi2_contingency,chi
results=[]
for col in cat_cols:
    table=pd.crosstab(df[col],df['churned'])
    chi2,p,dof,expected=chi2_contingency(table)
    results.append([col,p])
    chi_square_results=pd.DataFrame(results,columns=["Feature","P-value"])
chi_square_results['P-value']=chi_square_results['P-value'].round(6)
chi_square_results.sort_values("P-value")
```

```python
#cols with less than 0.05 value 
top_cat_cols=['churn_reason','plan_name','billing_address_city']
```

# Univariate Analysis

```python
for col in top_num_cols:
    fig,ax = plt.subplots(figsize=(6,4))
    sns.histplot(data=df, x=col, kde=True)
    ax.set_title(f"Distribution of {col}")
    plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="churn_reason")

plt.title("Distribution of Churn Reasons")
plt.xlabel("Churn Reason")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

# Bivariate Analysis

```python
for col in top_num_cols:
    fig,ax = plt.subplots(figsize=(6,4))
    sns.boxplot(data=df, x='churned',y=col)
    ax.set_title(f"churn vs {col}")
    plt.show()
```

## Final Insights

- Customers with a higher **days_since_last_order** are significantly more likely to churn, indicating that prolonged inactivity is one of the strongest indicators of customer churn.

- Customers with lower **NPS scores** tend to churn more frequently, suggesting that customer loyalty has a strong influence on retention.

- Churned customers generally raise more **support tickets** and have higher **complaint counts**, indicating that unresolved service issues contribute to customer attrition.

- Customers with lower **satisfaction scores** are more likely to churn, highlighting customer satisfaction as a key factor influencing retention.

- Churned customers show slightly higher **monthly spending**, **total spending**, and **average order value**, indicating that high-value customers can also churn if their expectations are not met.

- **Pages per session** exhibits a similar distribution for both churned and retained customers, suggesting that browsing activity alone is not a strong indicator of churn.

- **Tenure months** also shows a similar distribution across both customer groups, indicating that customer tenure has limited influence on churn in this dataset.

- **Predicted churn probability** clearly distinguishes churned from retained customers. However, this feature introduces **data leakage** and will be excluded during feature selection and model training.

## Multivariate Analysis

```python
plt.figure(figsize=(4,10))

corr = df[num_cols].corr()

sns.heatmap(
    corr[["churned"]].sort_values(by="churned", ascending=False),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation of Numerical Features with Churn")
plt.show()
```

## Correlation Analysis

### Observation

- **days_since_last_order** exhibits the strongest positive correlation with churn (**0.91**), indicating that customers who have not placed an order for a longer period are much more likely to churn.

- **predicted_churn_prob** has a very strong positive correlation (**0.90**) with the target variable. However, this feature introduces **data leakage** and will be excluded during feature selection and model training.

- **support_tickets_30d** shows a strong positive correlation (**0.83**), suggesting that customers with more support requests have a higher likelihood of churning.

- **complaint_count** has a moderate positive correlation (**0.57**), indicating that customers with more complaints are more likely to churn.

- **total_spend**, **avg_order_value**, and **monthly_spend** exhibit weak positive correlations (**≈0.29**), suggesting only a limited relationship with churn.

- **satisfaction_score** has a moderate negative correlation (**-0.41**), indicating that customers with higher satisfaction levels are less likely to churn.

- **nps_score** shows a strong negative correlation (**-0.87**), suggesting that highly loyal customers are significantly less likely to churn.

- Most of the remaining numerical features have correlation values close to **zero**, indicating a weak or negligible linear relationship with churn.


## Outlier Analysis 

```python
for col in top_num_cols:
    fig,ax = plt.subplots(figsize=(6,4))
    sns.boxplot(data=df, x=col)
    ax.set_title(f"{col}")
    plt.show()
```

### Outlier Analysis

- Boxplots were used to identify potential outliers in the numerical features.
- Most numerical features did not exhibit significant outliers.
- A few outliers were observed in **complaint_count**, where some customers recorded a higher number of complaints.
- Since **complaint_count** is a discrete count-based feature and the observed values are valid customer behavior rather than data entry errors, these observations were retained.
- Therefore, **no outlier treatment was performed**.
